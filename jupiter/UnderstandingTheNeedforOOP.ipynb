```python
# A class is  a blueprint for creating objects
# it defines attributes (data) and methods (behavior)
class Car:   # with the key word of class the interpreter will recogize that we are making some blue print
    def __init__(self, color):#this is like the constructor i think it become by writing the self
        self.color = color
        
    def start(self): # self describes the object for which the class is called
        print(f"{self.color} car started")
       

#first object
my_car = Car("red")
my_car.start()

#second object
mybrother_car = Car("red")
mybrother_car.start()

#Third car
myFather_car=Car("White")
myFather_car.start()

#Benefits of object oriented programming
#1.simplifies complex problems.
#2.Makes teamwork and collaboration easier
#3.Encourages reuse and scalability
#4.Forms the foundation of modern programming languages(Python,Java,C#,etc.)
```

```python
#Defining a class and creating objects/ template of a class
class Car:
    def drive(self):# in the self the object self will pass# self is that parameter which automatically gets passed
        print("The car is moving!")
        
```

```python
car1 = Car() # with this the compiler will understand that the car1 is an object that is made on the basis of Car template, its mean that all the attributes and methods that are defined in class will come insde this object
car1.drive()# if we write the car1 will come inside this method as self
```

```python
class Car:
    color = "red" # every object that will be made that will have a default red color, if we want we can change the color
    def drive(self):
        print(f"{self.color} car is moving!")
```

```python
car1 = Car()
car1.drive()
```

```python
class Car:
    color = "red"
    def drive(self):
        print(f"{self.color} car is moving!")
    def setColor(self, new_color):
        self.color = new_color
```

```python
car2 = Car()# we study this one earlier too that first write the name of your desird object
car2.setColor("green")# these methods are usually called setters
car2.drive()
```
