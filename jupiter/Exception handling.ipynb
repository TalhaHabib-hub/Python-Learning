```python
# Importance of Exceptions Handling
# Due to an error, Python will stop the program and display an error message: instead, we should handle the error and allow the program to continue running
# The program should not crash due to errors.
# example of errors -> liking putting invalid input,Dividing by zero also earlier we saw that the file which we want to open don't even exists
# if don't tell the python what to do when you find errors then python will crash the program when founds an error and on the other hand if we tell python what to do when you find error then python is not gonna crash Talha's program
# Exception Handling -> when an error occurs in your program, you inform python that the error was expected and that you are handling it.
# An unexpected error creates a negative impression on the client
```

```python
num = int(input("Enter a number: "))
print(10 / num) # python will tell if we put zero that for this code you have to handle this exception
```

```python
try: # try means run code like this if an unexpected error come (except) do this 
    num = int(input("Enter a number: "))
    print(10 / num )
except ZeroDivisionError: # these are exception types
    print("You can't divide by zero!")
except ValueError: # these are exception types
    print("That's not a number!")
except Exception as e: # these are exception types
    print(f"An Unexpected error occured: {e}")
```

```python
try:
    print("Running code...")
except:
    print("Error found!")
else: # the else will be executed when try code runs
    print("No errors found!")
finally: # The code which we will write in finally will be executed every time
    print("Done!")
```
