```python
#Variable Scope in Python
def show():
    x = 5 # local  , these have preference over the gloabal variables inside the function
    print(x)

x = 10  # global
show()
print(x)
```

```python
#lambda functions in Python
# A lambda in python is a small anonymous function written in a single line. use for simple, one-time tasks
add = lambda a, b: a + b

print(add(4,7))
```

```python
double_it = lambda a: a * 2

print(double_it(5))
```

```python
nums =[1,2,3,4]
squares =list(map(lambda x: x*x, nums))

print(squares)
```

```python
numbers = [1,2,3,4,5,6,7,8]
odd_numbers = list(filter(lambda x : x % 2 !=0, numbers))

print(odd_numbers)
```
