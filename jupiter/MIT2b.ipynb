```python
# break STATMENT
# immediately exits whatever loop it is in
# skips remaining expressions in code block
# exits only the innermost loop!
# just write break in the new statment
    
for n in range(3,13,2):
    print("Habib",n)
    for i in range (6):
        if i == 3:
            break
        print('Talha with',i)
    print("Hi")
    
    
```

```python
# so when to use for vs while loops
'''for loops
-> known numbers of iterations
-> can end early via break
-> uses a counter
-> can rewrite a for loop using a while loop but the vice versa isnot neccessary true'''

'''While loop
->unbounded number of iterations
->can end early via break
->can use a counter but must initializw befor loop and increment it inside loop
->may not be able to rewrite a while loop using a for loop'''
```
