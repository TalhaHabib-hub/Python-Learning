```python
# The  __init__()Constructor Method # the __init__ is a key word which we use to call the method exactly when an object is made , and this will assign value to the attributes for the perticular object (it is doing this in this case we can also perform other task with it) this  method of performing a task is called constructor method 
class Dog:
    def __init__(self):
        self.name = "Buddy"
    def bark(self):
        print(f"{self.name} says Woof!")
```

```python
dog1 = Dog() # this paranthesis are for constructor
dog1.bark()
```

```python
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color
```

```python
car1 = Car("Toyota","maroon")
print(car1.brand)
print(car1.color)
```

```python
car1.color = "Blue"
car2 = Car("Honda", "White")
print(car2.brand)
print(car2.color)
print(car1.color)
```

```python
#Insance Variables and instance Methods
class Student:
    def __init__(self, name, age, grade): #self shows for every particular object
        self.name = name #these three are instance variable# the self. with the attribute shows that these are instance variables
        self.age = age
        self.grade = grade

    def display_info(self):
        print(f"Student Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")

    def is_eligible(self):  
        if self.age >= 15:
            print(self.name, "is eligible for admission.")
        else:
            print(self.name, "is not eligible for admission.")
```

```python
#Encapsulation in Python -> Encapsulation means keeping all the data (attributes) and Methods of an object inside that object
# Private attributes are those attributes that can only be used within the class's own methods
# any data attribute that begins with two underscore will be a private attribute
class Student:
    def __init__(self, name, grade):
        self.__grade = grade #private attribute , this sort of data will only be used by the functions of the class and will not be accessible from the outside
# so to access such private data we use the methods of the class
student1 = Student("Talha", 90)
print(student1.__grade) #Error
```

```python
class Student:# the private data member of this class is access by torkhoychi war
    def __init__(self, name, grade):
        self.__grade = grade
    def get_grade(self):
        return self.__grade
student1 = Student("Talha", 99)
print(student1.get_grade())
```

```python
#using getter and setters in python
class Student:
    @property # why this buddy is here
    def grade(self):
        return self.__grade
    def __init__(self, name, grade):
        self.__grade = grade
    def get_grade(self):
        return self.__grade
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("invalid grade!")

student1 = Student("Ali", 89)# passing value to constructor
student1.set_grade(108)
print(student1.get_grade())
```

```python
#Inheritence in python-Parent&Child
#Inheritence allows a new class to reuse the properties and methods of and existing class.
class Animal:
    def speak(self):
        print("Some sound")
```
