```python
#Method Overriding
class Animal:
    def speak(self):
        print("Some sound")
```

```python
class Dog(Animal):
    def speak(self):# we are redefining here the speak method , so speak from Animal class will be overide
        print("woof! woof!")
```

```python
dog1 = Dog()
```

```python
dog1.speak()
```

```python
#Polymorphism in Python
#it means many form
#Polymorphism refers to the ability of a function or method to exist in multiple forms.
```

```python
class Dog:
    def speak(self):
        print("woof!")
```

```python
class Cat:
    def speak(self):
        print("Meow!")
```

```python
for pet in [Dog(), Cat()]: # what we have done here is we told Python first make a Dog object and then object for Cat and then place both inside pet
# pet is such a list of object inside which the first object is of Dog class and the second one is of Cat class
    pet.speak()
# in this loop we are iterating the whole list and we are  calling speak functioin on each object
```

```python
# Python Special Method
# Function -> A function can be used in the global context and operates independently of a class
# Method -> Method is that function that is defined within a class
# Special Function -> A special function defines built-in behavior automatically associated with certain operations.
```

```python
class Book:# we are wanting that people can use our class as string
    def __init__(self, title, author):
        self.title = title
        self.author = author
    def __str__(self):# default method of python language# with the __str__ function any object can use this class as a string(this is default method )not any user defined method
        return f"'{self.title}' by {self.author}"
```

```python
b = Book("1982", "George Orwell") # here is the object
```

```python
print(b) # now when we do this python will look that weather it can use the class as string then it finds yes and says you can because the python default string method is mentioned in the class
# in this way we can allow python to excute our class and print our class as string
```

```python
#another special function
# the number of songs will be the length
class Playlist:
    def __init__(self, songs):
        self.songs = songs
    def __len__(self): # default method
        return len(self.songs)
    def __add__(self, other): # default method
        return Playlist(self.songs + other.songs)
```

```python
p1 = Playlist(["Song1", "Song2",])
```

```python
p2 = Playlist(["song3"])
```

```python
print(len(p1)) #2
```

```python
print(len(p1 + p2)) #3 len is a default function in python
```

```python
#another method
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __str__(self):
        return f"({self.x}, {self.y})"
    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)
```

```python
p1 = Point(2, 3)
```

```python
p2 = Point(4, 1)
```

```python
print(p1 + p2)
```
