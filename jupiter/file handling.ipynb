```python
# Using Context Managers (with Statment) # so with is that context manager which make it possible to open the file till certain context
# python gives us a facility that by using context manager, we can tell this that  we are using the file for this much context and after that python should automatically close the file
# so we do this way
with open("janiiii.txt", "r") as file:
    data = file.read()
    print(data)
    # when the code under the with got executed the file will automatically get close and now we don't have to close it explicitly
# File close automatically here
# after with context manager their are alot more context managers
# open(): The standard file context manager for reading/writing files.
    
# tempfile.TemporaryFile()/ NamedTemporaryFile():# even gets deleted from hardisk
# Creates temporary files that delete themselves automatically
    
# zipfile.ZipFile(): Opens .zip files for reading/writing.

# contextlib.ExitStack: Lets you manage multiples files at once easily.
    
# pathlib.Path().open() # modern oop opening 
    
# Modern, object-oriented file opening via pathlib
```

```python
# Handling Exceptions During File Operation
# Common Issues in File Handling
# -> File not found
# -> Wrong path
# -> Wrong file mode
# due to which the program can crash for avoiding it we do exception handling
# An Exception is an error that occurs during the execution of a program, if we handle that exception python will not crash our program, and Talha if the program didn't handle the python will stop your program with giving error
# so lets do it on the editor
```

```python
with open("missing.txt", "r") as file:
    print(file.read())
```

```python
# as you saw there came an error when we tried to open the file which doesn't exist in the given path/directory
# Know we have to solve this problem
try: # try to open if not follow what is written in except
    with open("missing.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found!")
```

```python
# Talha with just printing file not found we can also do this " we can create the file" too
# for it lets go
filename = "my_info.txt"
name = "GitHub Copilot"
age = 1
```

```python
# Try to read existing file: if missing , create it with name and age, then read
try:
    with open(filename, "r") as fin: # here the file was not existing
        print(fin.read())
except FileNotFoundError: # if not then make open a file and gives us that as object as fout then we will give it a name and age
    with open(filename,"w") as fout:
        fout.write(f"{name}\n{age}\n")
    with open(filename, "r") as fin: # then the file is read and given to us and we have read it
        print(fin.read()) 
# so this way we can handle 
```
